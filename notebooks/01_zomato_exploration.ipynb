# Zomato Dataset — Data Exploration

## 1. Objective

The objective of this notebook is to understand the structure, quality, and characteristics of the historical Zomato restaurant dataset before performing data cleaning and feature engineering.

This dataset will be used as the historical business data component of the **Agentic AI-Based Business Potential Prediction in a Locality** project.

## 2. Project Context

The overall system aims to predict the potential of a food and beverage business in a given locality by combining:

- Historical business data from Zomato
- Current competition data from Google Places API
- Population and demographic data
- Locality and infrastructure data from OpenStreetMap (OSM)

In this notebook, we focus only on the **Zomato historical dataset**.

## 3. Objectives of Data Exploration

The following aspects of the dataset will be examined:

- Dataset dimensions (rows and columns)
- Column names and data types
- Missing values
- Duplicate records
- Unique values in categorical columns
- Distribution of important numerical variables
- Restaurant types and business categories
- Location distribution
- Cuisine distribution
- Rating distribution
- Cost distribution
- Potential features relevant to business performance prediction

## 4. Dataset Overview

The dataset contains information about restaurants in Bengaluru, including:

- Restaurant name and location
- Online ordering and table booking
- Restaurant ratings
- Votes/reviews
- Restaurant type
- Cuisines
- Approximate cost for two people
- Business category
- Listed city/location information
